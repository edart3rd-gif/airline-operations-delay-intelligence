```python
import pandas as pd

# Load the sample
flights = pd.read_csv("/content/flights_sample.csv")

# Start with a copy
df = flights.copy()

# Remove cancelled and diverted flights
df = df[
(df["CANCELLED"] == 0) &
(df["DIVERTED"] == 0)
].copy()

# Remove flights where arrival delay is unknown
df = df.dropna(subset=["ARRIVAL_DELAY"])

# Create target variable
# 1 = arrival delay > 15 minutes
# 0 = arrival delay <= 15 minutes
df["DELAYED"] = (df["ARRIVAL_DELAY"] > 15).astype(int)

# Fill missing delay-cause values with 0
cause_columns = [
"AIR_SYSTEM_DELAY",
"SECURITY_DELAY",
"AIRLINE_DELAY",
"LATE_AIRCRAFT_DELAY",
"WEATHER_DELAY"
]

df[cause_columns] = df[cause_columns].fillna(0)

print("Clean flights:", len(df))
print("Delayed flights:", df["DELAYED"].sum())
print("On-time flights:", (df["DELAYED"] == 0).sum())
print("Delay rate:", round(df["DELAYED"].mean() * 100, 2), "%")
```

```python
# Filter for United Airlines
ua = df[df["AIRLINE"] == "UA"].copy()

print("UNITED AIRLINES ANALYSIS")
print("------------------------")

print("United flights:", len(ua))
print("United delayed flights:", ua["DELAYED"].sum())
print("United delay rate:", round(ua["DELAYED"].mean() * 100, 2), "%")

print("Average arrival delay:",
round(ua["ARRIVAL_DELAY"].mean(), 2), "minutes")

print("Maximum arrival delay:",
round(ua["ARRIVAL_DELAY"].max(), 2), "minutes")
```

```python
# Compare airlines by delay performance

airline_summary = (
df.groupby("AIRLINE")
.agg(
flights=("DELAYED", "size"),
delayed_flights=("DELAYED", "sum"),
delay_rate=("DELAYED", "mean"),
avg_arrival_delay=("ARRIVAL_DELAY", "mean")
)
.reset_index()
)

# Convert delay rate to percentage
airline_summary["delay_rate"] = airline_summary["delay_rate"] * 100

# Sort by delay rate
airline_summary = airline_summary.sort_values(
"delay_rate",
ascending=False
)

print(airline_summary.round(2).to_string(index=False))
```

```python
# United airport performance

ua_airports = (
ua.groupby("ORIGIN_AIRPORT")
.agg(
flights=("DELAYED", "size"),
delayed_flights=("DELAYED", "sum"),
delay_rate=("DELAYED", "mean"),
avg_arrival_delay=("ARRIVAL_DELAY", "mean")
)
.reset_index()
)

# Convert to percentage
ua_airports["delay_rate"] = ua_airports["delay_rate"] * 100

# Only include airports with at least 30 United flights
ua_airports = ua_airports[
ua_airports["flights"] >= 30
].copy()

# Sort by highest delay rate
ua_airports = ua_airports.sort_values(
"delay_rate",
ascending=False
)

print("UNITED AIRPORT PERFORMANCE")
print("--------------------------")

print(
ua_airports.head(10).round(2).to_string(index=False)
)
```

```python
# Analyze causes of United delays

delay_causes = {
    "Air System": "AIR_SYSTEM_DELAY",
    "Security": "SECURITY_DELAY",
    "Airline": "AIRLINE_DELAY",
    "Late Aircraft": "LATE_AIRCRAFT_DELAY",
    "Weather": "WEATHER_DELAY"
}

cause_results = []

for cause_name, column in delay_causes.items():
    total_minutes = ua[column].sum()
    affected_flights = (ua[column] > 0).sum()

    cause_results.append({
        "Cause": cause_name,
        "Total Delay Minutes": total_minutes,
        "Flights Affected": affected_flights
    })

# Create summary DataFrame
cause_summary = pd.DataFrame(cause_results)

# Sort by total delay minutes
cause_summary = cause_summary.sort_values(
    by="Total Delay Minutes",
    ascending=False
)

# Reset index
cause_summary = cause_summary.reset_index(drop=True)

# Display results
print("UNITED DELAY CAUSES")
print("-------------------")
print(cause_summary.to_string(index=False))
```

```python
import matplotlib.pyplot as plt

# Sort causes from highest to lowest
plot_data = cause_summary.sort_values(
"Total Delay Minutes",
ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
plot_data["Cause"],
plot_data["Total Delay Minutes"]
)

plt.xlabel("Total Delay Minutes")
plt.ylabel("Delay Cause")
plt.title("United Airlines — Total Delay Minutes by Cause")

plt.tight_layout()
plt.show()
```

```python
# Create a route identifier
ua["ROUTE"] = (
ua["ORIGIN_AIRPORT"] + " → " + ua["DESTINATION_AIRPORT"]
)

# Analyze United routes
ua_routes = (
ua.groupby("ROUTE")
.agg(
flights=("DELAYED", "size"),
delayed_flights=("DELAYED", "sum"),
delay_rate=("DELAYED", "mean"),
avg_arrival_delay=("ARRIVAL_DELAY", "mean")
)
.reset_index()
)

# Convert delay rate to percentage
ua_routes["delay_rate"] = ua_routes["delay_rate"] * 100

# Only analyze routes with at least 20 flights
ua_routes = ua_routes[
ua_routes["flights"] >= 20
].copy()

# Sort by delay rate
ua_routes = ua_routes.sort_values(
"delay_rate",
ascending=False
)

print("UNITED WORST ROUTES")
print("--------------------")

print(
ua_routes.head(10).round(2).to_string(index=False)
)
```

```python
import matplotlib.pyplot as plt

route_plot = ua_routes.head(10).sort_values(
"delay_rate",
ascending=True
)

plt.figure(figsize=(10, 6))

plt.barh(
route_plot["ROUTE"],
route_plot["delay_rate"]
)

plt.xlabel("Delay Rate (%)")
plt.ylabel("Route")
plt.title("United Airlines — Worst Routes by Delay Rate")

plt.tight_layout()
plt.show()
```

```python
import sqlite3
import pandas as pd

# Create an in-memory SQL database
conn = sqlite3.connect(":memory:")

# Load our cleaned flight data into SQL
df.to_sql(
"flights",
conn,
index=False,
if_exists="replace"
)

# SQL analysis: airline performance
query = """
SELECT
AIRLINE,
COUNT(*) AS flights,
SUM(CASE WHEN ARRIVAL_DELAY > 15 THEN 1 ELSE 0 END) AS delayed_flights,
ROUND(
100.0 * SUM(CASE WHEN ARRIVAL_DELAY > 15 THEN 1 ELSE 0 END) / COUNT(*),
2
) AS delay_rate,
ROUND(AVG(ARRIVAL_DELAY), 2) AS avg_arrival_delay
FROM flights
GROUP BY AIRLINE
ORDER BY delay_rate DESC;
"""

sql_results = pd.read_sql_query(query, conn)

print("SQL AIRLINE PERFORMANCE")
print("-----------------------")
print(sql_results.to_string(index=False))
```
